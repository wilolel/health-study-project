I del 1 användes en boxplot för att undersökte åldersfördelning hos rökare och icke-rökare. Båda grupperna hade en medianålder på 50 år och en snarlik åldersfördelning. 

För att titta vidare på detta skapades ett grupperat stapeldiagram som visar hur andelen rökare varierar mellan åldersgrupper. Beräkningen av procentandelarna och visualiseringen delades upp i två separata funktioner. Eftersom åldersgrupperna innehåller olika många personer visas andelarna i procent iställer för antal.

Diagrammet visar att andelen rökare är relativt likartad bland personer mellan åldrarna 30 och 79, där den ligger på omkring 24-29%. Den yngsta åldersgruppen, 18-29 år, har en lägre andel rökare på ungefär 18%. Den äldsta åldersgruppen, 80-90 år, sticker ut med ungefär 42% rökare. Resultatet bör dock tolkas försiktigt eftersom procentandelar kan påverkas mycket av om gruppen innehåller få personer.

Jämfört med boxploten från del 1 ger det nya diagrammet en mer detaljerad bild av hur rökning förekommer i olika åldrar. Även om rökare och icke-rökare har samma medianålder innebär det inte att andelen rökare är densamma i alla åldersgrupper.
