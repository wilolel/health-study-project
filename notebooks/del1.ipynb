## 1.1 Statistiska värden

```python
# Räkna medel, median, min, max för age, weight, height, systolic_bp, cholesterol
# Skapa minst 3 grafer, t.ex. histogram över blodtryck, boxplot vikt per kön, stapeldiagram för andel rökare
# VG - motivera metodval i markdown och hänvisa till dokumentation eller annan källa

# Räknar ut medel, median, min och max för valfri kolumn i datasetet
def descriptive_values(df, column):
    return {
        "mean": df[column].mean(),
        "median": df[column].median(),
        "min": df[column].min(),
        "max": df[column].max()
    }

columns = ["age", "weight", "height", "systolic_bp", "cholesterol"]

results = {}

for column in columns:
    results[column] = descriptive_values(df, column)

pd.DataFrame(results).T
```

## 1.2. Grafer
### 1.2.1 Boxplot - jämför ålder mellan rökare och icke-rökare

```python
smokers = df.loc[df["smoker"] == "Yes", "age"]
non_smokers = df.loc[df["smoker"] == "No", "age"]
fig, ax = plt.subplots(figsize=(9, 6))

box = ax.boxplot([smokers, non_smokers], tick_labels=["Smokers", "Non-smokers"], patch_artist=True)
box["boxes"][0].set_facecolor("tab:pink")
box["boxes"][1].set_facecolor("tab:green")

ax.set_title("Smokers and non-smokers - Age distribution")
ax.set_ylabel("Age")
ax.grid(True, axis="y", alpha=0.3)
ax.axhline(50, color="black", linestyle="--")
ax.annotate("Median age: 50 for both groups", 
            xy=(1.5, 50), 
            xytext=(1.4, 65),
            arrowprops={"arrowstyle": "->"}
)

for median in box["medians"]:
    median.set_color("none")

plt.tight_layout()
plt.show()
```

### 1.2.2 Stapeldiagram - sjukdom och rökning

```python
smoker_disease = df.loc[(df["smoker"] == "Yes") & (df["disease"] == 1)]
smoker_healthy = df.loc[(df["smoker"] == "Yes") & (df["disease"] == 0)]

non_smoker_disease = df.loc[(df["smoker"] == "No") & (df["disease"] == 1)]
non_smoker_healthy = df.loc[(df["smoker"] == "No") & (df["disease"] == 0)]

total_smokers = (df["smoker"] == "Yes").sum()
total_non_smokers = (df["smoker"] == "No").sum()

smoker_disease_pct = len(smoker_disease) / total_smokers * 100
smoker_healthy_pct = len(smoker_healthy) / total_smokers * 100

non_smoker_disease_pct = len(non_smoker_disease) / total_non_smokers * 100
non_smoker_healthy_pct = len(non_smoker_healthy) / total_non_smokers * 100

groups = ["Smokers", "Non-smokers"]
healthy = [smoker_healthy_pct, non_smoker_healthy_pct]
disease = [smoker_disease_pct, non_smoker_disease_pct]

fig, ax = plt.subplots(figsize=(9, 6))

ax.bar(groups, disease, label="Disease", color="tab:pink")
ax.bar(groups, healthy, bottom=disease, label="Healthy", color="tab:green")

ax.grid(True, axis="y", alpha=0.3)
ax.set_ylim(0, 100)
ax.set_yticks(range(0, 101, 10))
ax.set_title("Disease among smokers and non-smokers")
ax.set_ylabel("Percentage")

ax.legend()

plt.show()
```
